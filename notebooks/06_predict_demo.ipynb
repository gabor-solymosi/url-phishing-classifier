```python
from pathlib import Path
import os
import sys

# Import from the repo root; the model itself loads on the first prediction.
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
os.chdir(ROOT)
sys.path.insert(0, str(ROOT))
import pandas as pd
from predict import predict

examples = [
    "https://www.PayPal-secure.login.tk/webscr?cmd=x",
    "https://www.google.com/",
    "",
    None,
    12345,
    "http://[::1",
    "https://例え.テスト/パス",
    "javascript:alert(1)",
]
results = [predict(value) for value in examples]
display(pd.DataFrame([
    {"input": repr(value), **result}
    for value, result in zip(examples, results)
]))
```
